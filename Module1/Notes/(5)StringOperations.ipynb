## String Operations

### String
 - A sequence of characters (e.g. "Theo Ermino")
 - either single or double quotes are acceptable
 - can be spaces or digits (e.g "   ", "23434")
 - also can be special characters (e.g. #$%^&*)


### String as an Ordered Sequence
 - Each element in the sequence can be accessed using an index represented by the array of numbers
 - can use negative indexing
 - can bind assign string to another variable
