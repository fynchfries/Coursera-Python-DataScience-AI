## Python Data Types

### Data Types
 - it how python represents different types of data.

### Data Types: Over View:
 - can be numbers or real numbers (e.g. 1, 2.3)
 - can be words (e.g. vowels)

### Example
  Expression | Data Type
    11          int
    2.3         float
  "Hello"       str

### Data type: int
 - can be positive
 - can be negative

### Data type: float
 - real numbers
 - include the integers and number between the integers (e.g. x = 5: 5.0)

### Changing Expression Type
 `Type Casting` converting an expression (e.g. float(3):3.0; int(3.99):3)

### Data Types: Boolean
 - contain 2 values
  > True
  > False
 - bool short for boolean
