## About Jupyter

 - Enables creation and sharing of documents containing equations, live coding, visualizations, and narrative text.

## Why Jupter?
 - Popularity
 - Ease of use, regardless of your expertise
 - Interactive interface

## Key Features of Jupyter:
 - Write and Execute code interactively
 - Support various programming language through its kernel system.
 - Reach media integration, allowing users to generate interactive plots, charts, images, videos, and more directly within the document.
 - Integrates with popular libraries and framework used in Data Science.
 - Sharable.
